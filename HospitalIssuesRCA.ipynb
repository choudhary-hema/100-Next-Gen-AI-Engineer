```python
from langchain_google_genai import ChatGoogleGenerativeAI
import os
import getpass
```

```python
os.environ["GEMINI_API_KEY"]=getpass.getpass("enter your api key here")
llm=ChatGoogleGenerativeAI(
    model='gemini-3.6-flash'
)
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
```

```python
hospital_issues_prompt=PromptTemplate(
  template="""
    Act as a Healthcare Operations Consultant. Analyze the hospital data/context
    below to identify key issues and root causes across three areas:
      - Patient Experience (wait times, communication, care/discharge quality)
      - Operational Efficiency (bottlenecks, bed turnover, staffing/workflows)
      - Financial Performance (revenue leakage, claim denials, excess costs)

   List the issue in bullet points
  """
)
```

```python
chain1=hospital_issues_prompt | llm | StrOutputParser()
```

```python
issues=chain1.invoke({})

print(issues)
```

```python
rca_prompt=PromptTemplate(
  input_variables=["issues"],
  template="""
    You are a Senior Troubleshooting & Root Cause Analysis (RCA) Engineer.
    Your task is to investigate the reported issues, separate symptoms from underlying causes, and determine the true root cause(s).

    issues: {issues}
  """
)
```

```python
chain2=rca_prompt | llm | StrOutputParser
```

```python
solutions=chain2.invoke({
    "issues":issues
})
print(solutions)
```

```python
report_prompt=PromptTemplate(
    input_variables=["solutions"],
    template="""
    Prepare a concise Incident & RCA Report from the findings covering Executive Summary,
    Issues Reported, Root Cause, Immediate Fixes, and Preventive Action Items.
    Stay factual and do not invent missing details.
    solutions: {solutions}
    """
)
```

```python
chain3=report_prompt | llm | StrOutputParser()
```

```python
report=chain3.invoke({
    "solutions":solutions
})
print(report)
```
