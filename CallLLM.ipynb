```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_excel("/content/My_Financial_Report_Dataset.xlsx")
```

```python
data.head()
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
```

```python
kpi_prompt=PromptTemplate(
    input_variables=["data"],
    template="""
        You are financial Analyst and you have been provided with financial data.
        Extract the most important financial and business KPIs from the data and
        present them in bullet point format.

        financial data:{data}"""
)
```

```python
chain1=kpi_prompt | llm | StrOutputParser()
```

```python
kpis=chain1.invoke(
    {"data":data}
)
print(kpis)
```

```python
benchmark_prompt=PromptTemplate(
    input_variables=["kpis"],
    template="""
      You are senior Strategy consultant and your job is to critically analyse the kpis shared and
      establish appropriaste performance for consultant evaluation.

      For every major KPI provide the following
      - Define a reasonable target or benchmark
      - State the current performance
      - Classify the performance as
       - Strong
       - Moderate
       - Needs Improvements

      kpis:{kpis}
    """
)
```

```python
chain2=benchmark_prompt | llm | StrOutputParser()
```

```python
benchmark = chain2.invoke({
    "kpis":kpis
})
```

```python
print(benchmark)
```

```python
executive_prompt=PromptTemplate(
    input_variables=["benchmark"],
    template="""
      You are a executive and you have been provided with
      a benchmark data. Extract the benchmark information and generate the executive report.
      benchmark:{benchmark}
    """
)
```

```python
chain3=executive_prompt | llm | StrOutputParser()
```

```python
executive_report=chain3.invoke({
    "benchmark":benchmark
})
```

```python
print(executive_report)
```
